# Hemp OS API Example: Exporting Knowledge Graph & Vectors

```python
import requests
import networkx as nx
import matplotlib.pyplot as plt

HEMP_OS_API_URL = 'http://localhost:3000/api'
HEADERS = {'x-api-key': 'your_secret_password'}
```

```python
# Fetch Graph
res = requests.get(f'{HEMP_OS_API_URL}/graph', headers=HEADERS)
graph_data = res.json()
nodes = graph_data['nodes']
edges = graph_data['edges']

G = nx.DiGraph()
for n in nodes:
    G.add_node(n['id'], label=n.get('label', n['id']))
for e in edges:
    G.add_edge(e['source'], e['target'], relation=e['relation'])

print(f'Graph has {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.')

# Optional: Simple Plot
plt.figure(figsize=(10,10))
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color='lightgreen', edge_color='gray')
plt.show()
```
