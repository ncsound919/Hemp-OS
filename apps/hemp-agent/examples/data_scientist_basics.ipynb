# Hemp OS API Example: Pulling Studies & Omics Data

This notebook demonstrates how a Data Scientist can programmatically query Hemp OS.

```python
import requests
import pandas as pd

HEMP_OS_API_URL = 'http://localhost:3000/api'
# Remember to pass API key if configured
HEADERS = {'x-api-key': 'your_secret_password'}
```

## 1. Fetching Clinical Studies

```python
res = requests.get(f'{HEMP_OS_API_URL}/db/studies', headers=HEADERS)
studies = res.json()
df_studies = pd.DataFrame(studies)
display(df_studies.head())
```

## 2. Fetching Omics Signatures

```python
res_omics = requests.get(f'{HEMP_OS_API_URL}/db/omics', headers=HEADERS)
omics = res_omics.json()
df_omics = pd.DataFrame(omics)
display(df_omics.head())
```
